**Шаг 1. Загрузка и подготовка CIFAR-10.**

```python
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F

from torch.utils.data import Dataset, DataLoader
from torchvision import datasets, transforms, models

import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    ConfusionMatrixDisplay
)

import time
from copy import deepcopy
from PIL import Image

torch.manual_seed(42)
np.random.seed(42)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Устройство:", device)
print("PyTorch:", torch.__version__)
```

Трансформации и загрузка CIFAR-10

```python
# Преобразования для обучающей выборки
train_transform = transforms.Compose([
    transforms.Resize(64),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(15),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.ToTensor(),
    transforms.Normalize(
        (0.5, 0.5, 0.5),
        (0.5, 0.5, 0.5)
    )
])

# Преобразования для валидационной выборки
val_transform = transforms.Compose([
    transforms.Resize(64),
    transforms.ToTensor(),
    transforms.Normalize(
        (0.5, 0.5, 0.5),
        (0.5, 0.5, 0.5)
    )
])

print("Трансформации созданы успешно!")
```

Фильтрация классов

```python
CAT_IDX, DOG_IDX = 3, 5

class BinaryCIFAR(Dataset):
    """Оставляет только кошек и собак."""

    def __init__(self, dataset, transform):
        self.transform = transform
        self.images = []
        self.labels = []

        for img, label in dataset:
            if label == CAT_IDX:
                self.images.append(img)
                self.labels.append(0)

            elif label == DOG_IDX:
                self.images.append(img)
                self.labels.append(1)

    def __len__(self):
        return len(self.images)

    def __getitem__(self, idx):
        img = self.images[idx]
        label = self.labels[idx]

        return self.transform(img), label


# Загружаем CIFAR-10
train_full = datasets.CIFAR10(
    root="./data",
    train=True,
    download=True
)

val_full = datasets.CIFAR10(
    root="./data",
    train=False,
    download=True
)

# Создаём выборки из двух классов
train_dataset = BinaryCIFAR(
    train_full,
    transform=train_transform
)

val_dataset = BinaryCIFAR(
    val_full,
    transform=val_transform
)

print("Train size:", len(train_dataset))
print("Val size:", len(val_dataset))
print("Класс 0: кошка")
print("Класс 1: собака")
```

Создание DataLoader

```python
train_loader = DataLoader(
    train_dataset,
    batch_size=64,
    shuffle=True,
    num_workers=2
)

val_loader = DataLoader(
    val_dataset,
    batch_size=128,
    shuffle=False,
    num_workers=2
)

print("Train batches:", len(train_loader))
print("Val batches:", len(val_loader))
```

Визуализация изображений

**Шаг 2. Построение архитектуры SimpleCNN**

```python
# Функция создания одного свёрточного блока
def conv_block(in_channels, out_channels):
    return nn.Sequential(
        nn.Conv2d(
            in_channels,
            out_channels,
            kernel_size=3,
            padding=1
        ),
        nn.BatchNorm2d(out_channels),
        nn.ReLU(inplace=True),
        nn.MaxPool2d(kernel_size=2, stride=2)
    )


# Архитектура SimpleCNN
class SimpleCNN(nn.Module):
    def __init__(self, num_classes=1):
        super().__init__()

        self.features = nn.Sequential(
            conv_block(3, 32),
            conv_block(32, 64),
            conv_block(64, 128),
            conv_block(128, 256)
        )

        self.classifier = nn.Sequential(
            nn.AdaptiveAvgPool2d(1),
            nn.Flatten(),
            nn.Dropout(0.5),
            nn.Linear(256, num_classes)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x


# Создание модели
model = SimpleCNN(num_classes=1).to(device)

# Общее количество обучаемых параметров
total_params = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

# Количество параметров в первых двух свёрточных слоях
first_conv = model.features[0][0]
second_conv = model.features[1][0]

print("Архитектура SimpleCNN создана!")
print("Устройство:", device)

print(
    "Параметры первого Conv2d:",
    sum(p.numel() for p in first_conv.parameters())
)

print(
    "Параметры второго Conv2d:",
    sum(p.numel() for p in second_conv.parameters())
)

print(f"Всего обучаемых параметров: {total_params:,}")
```

**Шаг 3. Функция потерь и оптимизатор**

```python
# Функция потерь
criterion = nn.BCEWithLogitsLoss()

# Оптимизатор
optimizer = optim.AdamW(
    model.parameters(),
    lr=1e-3,
    weight_decay=1e-2
)

# Планировщик скорости обучения
scheduler = optim.lr_scheduler.CosineAnnealingLR(
    optimizer,
    T_max=15
)

print("Функция потерь:", criterion)
print("Оптимизатор:", optimizer.__class__.__name__)
print("Начальный Learning Rate:", optimizer.param_groups[0]["lr"])
print("Планировщик:", scheduler.__class__.__name__)
```

Проверка модели на одном batch

```python
# Получаем один batch
X_batch, y_batch = next(iter(train_loader))

X_batch = X_batch.to(device)
y_batch = y_batch.float().unsqueeze(1).to(device)

# Прямой проход
logits = model(X_batch)

# Вычисление функции потерь
loss = criterion(logits, y_batch)

print("Размер входных данных:", X_batch.shape)
print("Размер меток:", y_batch.shape)
print("Размер выхода модели:", logits.shape)
print("Значение Loss:", loss.item())
```

**Шаг 4. Функции обучения и оценки модели**

```python
def train_epoch(model, loader, criterion, optimizer, device):
    model.train()

    running_loss = 0.0
    correct = 0
    total = 0

    for X_batch, y_batch in loader:
        X_batch = X_batch.to(device)
        y_batch = y_batch.float().unsqueeze(1).to(device)

        optimizer.zero_grad()

        logits = model(X_batch)
        loss = criterion(logits, y_batch)

        loss.backward()
        optimizer.step()

        running_loss += loss.item() * X_batch.size(0)

        preds = (torch.sigmoid(logits) > 0.5).float()

        correct += (preds == y_batch).sum().item()
        total += y_batch.size(0)

    return running_loss / total, correct / total


@torch.no_grad()
def evaluate(model, loader, criterion, device):
    model.eval()

    running_loss = 0.0
    correct = 0
    total = 0

    all_probs = []
    all_labels = []

    for X_batch, y_batch in loader:
        X_batch = X_batch.to(device)
        y_batch = y_batch.float().unsqueeze(1).to(device)

        logits = model(X_batch)
        loss = criterion(logits, y_batch)

        running_loss += loss.item() * X_batch.size(0)

        probs = torch.sigmoid(logits)
        preds = (probs > 0.5).float()

        correct += (preds == y_batch).sum().item()
        total += y_batch.size(0)

        all_probs.extend(probs.cpu().numpy())
        all_labels.extend(y_batch.cpu().numpy())

    return (
        running_loss / total,
        correct / total,
        np.array(all_probs),
        np.array(all_labels)
    )


def fit(
    model,
    train_loader,
    val_loader,
    criterion,
    optimizer,
    scheduler,
    epochs=15,
    patience=5,
    device=device,
    verbose=True
):
    history = {
        "train_loss": [],
        "val_loss": [],
        "train_acc": [],
        "val_acc": []
    }

    best_val_loss = float("inf")
    best_weights = deepcopy(model.state_dict())
    epochs_no_improve = 0

    for epoch in range(1, epochs + 1):
        t0 = time.time()

        train_loss, train_acc = train_epoch(
            model, train_loader, criterion, optimizer, device
        )

        val_loss, val_acc, _, _ = evaluate(
            model, val_loader, criterion, device
        )

        scheduler.step()

        dt = time.time() - t0

        history["train_loss"].append(train_loss)
        history["val_loss"].append(val_loss)
        history["train_acc"].append(train_acc)
        history["val_acc"].append(val_acc)

        if val_loss < best_val_loss:
            best_val_loss = val_loss
            best_weights = deepcopy(model.state_dict())
            epochs_no_improve = 0
        else:
            epochs_no_improve += 1

        if verbose:
            print(
                f"Epoch {epoch:3d} | {dt:.1f}s | "
                f"Train loss {train_loss:.4f} acc {train_acc:.4f} | "
                f"Val loss {val_loss:.4f} acc {val_acc:.4f}"
            )

        if epochs_no_improve >= patience:
            print(
                f"Early stopping at epoch {epoch} "
                f"(best val loss {best_val_loss:.4f})"
            )
            break

    model.load_state_dict(best_weights)

    return history


def plot_history(history, title=""):
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))

    axes[0].plot(history["train_loss"], label="train")
    axes[0].plot(history["val_loss"], label="val")
    axes[0].set_xlabel("Epoch")
    axes[0].set_ylabel("Loss")
    axes[0].set_title(f"{title} — Loss")
    axes[0].legend()

    axes[1].plot(history["train_acc"], label="train")
    axes[1].plot(history["val_acc"], label="val")
    axes[1].set_xlabel("Epoch")
    axes[1].set_ylabel("Accuracy")
    axes[1].set_title(f"{title} — Accuracy")
    axes[1].legend()

    plt.tight_layout()
    plt.show()


print("Функции обучения и оценки созданы успешно!")
```

**Шаг 5. Обучение SimpleCNN**

```python
history = fit(
    model,
    train_loader,
    val_loader,
    criterion,
    optimizer,
    scheduler,
    epochs=15,
    patience=5
)

plot_history(history, "SimpleCNN")
```

**Шаг 6. Transfer Learning с ResNet-18**

Шаг 6.1. Создание ResNet-18

```python
# Загружаем ResNet-18 с предобученными весами
resnet = models.resnet18(
    weights=models.ResNet18_Weights.IMAGENET1K_V1
)

# Заменяем последний слой для бинарной классификации
resnet.fc = nn.Linear(512, 1)

# Переносим модель на GPU
resnet = resnet.to(device)

# Подсчёт параметров
resnet_params = sum(
    p.numel() for p in resnet.parameters()
)

print("ResNet-18 успешно создана!")
print(f"Количество параметров: {resnet_params:,}")
print("Устройство:", device)
```

Шаг 6.2. Настройка Fine-tuning

```python
# Разрешаем обучение всех параметров
for param in resnet.parameters():
    param.requires_grad = True

# Функция потерь
criterion_resnet = nn.BCEWithLogitsLoss()

# Оптимизатор с меньшей скоростью обучения
optimizer_resnet = optim.AdamW(
    resnet.parameters(),
    lr=1e-4,
    weight_decay=1e-2
)

# Планировщик
scheduler_resnet = optim.lr_scheduler.CosineAnnealingLR(
    optimizer_resnet,
    T_max=10
)

print("Fine-tuning настроен!")
print(
    "Обучаемых параметров:",
    sum(p.numel() for p in resnet.parameters() if p.requires_grad)
)
```

Шаг 6.3. Обучение ResNet-18

```python
start_time = time.time()

history_resnet = fit(
    resnet,
    train_loader,
    val_loader,
    criterion_resnet,
    optimizer_resnet,
    scheduler_resnet,
    epochs=10,
    patience=5
)

resnet_training_time = time.time() - start_time

plot_history(history_resnet, "ResNet-18")

print(
    f"Время обучения ResNet-18: "
    f"{resnet_training_time:.2f} секунд"
)
```

Таким образом, Feature Maps позволяют увидеть, как нейронная сеть преобразует исходное изображение и выделяет признаки, необходимые для последующей классификации кошек и собак.

Шаг 9. Grad-CAM — визуализация внимания модели

```python
import matplotlib.cm as cm

class GradCAM:
    def __init__(self, model, target_layer):
        self.model = model
        self.gradients = None
        self.activations = None

        def forward_hook(module, inputs, output):
            self.activations = output.detach()

            output.register_hook(
                lambda grad: setattr(
                    self, "gradients", grad.detach()
                )
            )

        self.handle = target_layer.register_forward_hook(
            forward_hook
        )

    def generate(self, input_tensor):
        self.model.eval()
        self.model.zero_grad(set_to_none=True)

        output = self.model(input_tensor)

        probability = torch.sigmoid(output).item()
        predicted_class = int(probability >= 0.5)

        # Объясняем именно предсказанный класс
        score = output[0, 0] if predicted_class == 1 else -output[0, 0]

        score.backward()

        weights = self.gradients.mean(
            dim=(2, 3), keepdim=True
        )

        cam = (weights * self.activations).sum(
            dim=1, keepdim=True
        )

        cam = F.relu(cam)
        cam = cam - cam.min()
        cam = cam / (cam.max() + 1e-8)

        return cam[0, 0].cpu().numpy(), predicted_class, probability


# Последний свёрточный слой SimpleCNN
target_layer = model.features[3][0]

grad_cam = GradCAM(model, target_layer)

class_names = ["Cat", "Dog"]

# Выбираем 5 случайных изображений
indices = np.random.choice(
    len(val_dataset), 5, replace=False
)

fig, axes = plt.subplots(5, 3, figsize=(12, 18))

for row, idx in enumerate(indices):
    img, label = val_dataset[idx]

    input_tensor = img.unsqueeze(0).to(device)

    # Строим Grad-CAM
    cam, prediction, probability = grad_cam.generate(
        input_tensor
    )

    # Восстанавливаем исходное изображение
    img_show = img.permute(1, 2, 0).numpy()
    img_show = img_show * 0.5 + 0.5
    img_show = np.clip(img_show, 0, 1)

    # Изменяем размер тепловой карты
    cam_resized = np.array(
        Image.fromarray(
            (cam * 255).astype(np.uint8)
        ).resize((64, 64))
    ) / 255.0

    heatmap = cm.jet(cam_resized)[:, :, :3]

    # Накладываем карту на изображение
    overlay = np.clip(
        0.5 * img_show + 0.5 * heatmap,
        0, 1
    )

    axes[row, 0].imshow(img_show)
    axes[row, 0].set_title(
        f"Оригинал: {class_names[label]}"
    )

    axes[row, 1].imshow(cam, cmap="jet")
    axes[row, 1].set_title("Grad-CAM")

    axes[row, 2].imshow(overlay)
    axes[row, 2].set_title(
        f"Предсказание: {class_names[prediction]}\n"
        f"Вероятность Dog: {probability:.2f}"
    )

    for ax in axes[row]:
        ax.axis("off")

plt.suptitle("Grad-CAM: области внимания SimpleCNN", fontsize=14)
plt.tight_layout()
plt.show()
```

Задание 9.1. Визуализация и анализ Grad-CAM

Для пяти изображений была построена визуализация Grad-CAM, позволяющая определить области, которые оказывают наибольшее влияние на решения нейронной сети SimpleCNN.

По результатам эксперимента все пять изображений были правильно классифицированы. При анализе тепловых карт видно, что модель обращает внимание на различные части животных. На изображениях собак наиболее активные области располагаются преимущественно в районе головы и морды. На изображениях кошек наблюдаются активации в области головы, тела и шерсти.

При этом на отдельных изображениях активные области частично захватывают окружающий фон или передний план. Это означает, что модель может использовать не только характерные особенности самого животного, но и дополнительные визуальные признаки изображения.

Полученные результаты частично согласуются с тем, как человек различает кошек и собак: при классификации важную роль играют форма головы и морды, уши, пропорции тела и текстура шерсти. Однако нельзя утверждать, что нейронная сеть распознаёт именно эти анатомические признаки: Grad-CAM показывает области влияния на предсказание, а не конкретные объекты или части тела.

Таким образом, визуализация Grad-CAM позволила оценить механизм принятия решений нейронной сетью и показала, что модель преимущественно использует информативные области изображений животных, хотя в отдельных случаях может учитывать и окружающий фон.

```python
# Выбираем изображение именно кошки или собаки
np.random.seed(42)
idx = np.random.randint(len(train_dataset))

img_pil = train_dataset.images[idx]
label = train_dataset.labels[idx]

augmentations = [
    ('Оригинал', transforms.Compose([
        transforms.Resize(64),
        transforms.ToTensor()
    ])),
    ('Поворот +15°', transforms.Compose([
        transforms.Resize(64),
        transforms.RandomRotation((15, 15)),
        transforms.ToTensor()
    ])),
    ('Поворот -15°', transforms.Compose([
        transforms.Resize(64),
        transforms.RandomRotation((-15, -15)),
        transforms.ToTensor()
    ])),
    ('Отражение', transforms.Compose([
        transforms.Resize(64),
        transforms.RandomHorizontalFlip(p=1.0),
        transforms.ToTensor()
    ])),
    ('Яркость +', transforms.Compose([
        transforms.Resize(64),
        transforms.ToTensor(),
        transforms.Lambda(lambda x: (x + 0.3).clamp(0, 1))
    ])),
    ('Яркость -', transforms.Compose([
        transforms.Resize(64),
        transforms.ToTensor(),
        transforms.Lambda(lambda x: (x - 0.3).clamp(0, 1))
    ]))
]

fig, axes = plt.subplots(1, 6, figsize=(18, 4))

for i, (name, aug) in enumerate(augmentations):
    img_aug = aug(img_pil)

    axes[i].imshow(img_aug.permute(1, 2, 0).numpy())
    axes[i].set_title(name, fontsize=10)
    axes[i].axis('off')

plt.suptitle(
    f'Data Augmentation: {"Cat" if label == 0 else "Dog"}',
    fontsize=14
)

plt.tight_layout()
plt.show()

print("Исходный класс:", "Cat" if label == 0 else "Dog")
```

Задание 10.1. Какие преобразования безопасны для кошек и собак?


Для классификации изображений кошек и собак безопасными являются небольшие повороты, горизонтальное отражение, изменение яркости и контрастности. Такие преобразования изменяют внешний вид изображения, но сохраняют его основной смысл и принадлежность объекта к исходному классу.

Повороты позволяют модели лучше распознавать животных под разными углами. Горизонтальное отражение помогает учитывать различное расположение объектов. Изменение яркости и контрастности повышает устойчивость модели к разным условиям освещения.

Горизонтальное отражение допустимо для кошек и собак, поскольку при изменении направления изображения животное остаётся представителем того же класса. Однако для цифр и букв такое преобразование может изменить смысл символа: например, отражённая цифра 6 может напоминать 9, а некоторые буквы могут стать неузнаваемыми.

Таким образом, Data Augmentation увеличивает разнообразие обучающих примеров, помогает уменьшить переобучение и улучшает способность нейронной сети распознавать объекты в различных условиях.

Accuracy показывает общую долю правильных предсказаний, однако эта метрика не позволяет полностью оценить качество классификатора. F1-score учитывает одновременно Precision и Recall, а AUC характеризует способность модели различать два класса при различных порогах классификации.

Precision важнее в ситуациях, когда необходимо минимизировать ложное определение кошек как собак. Например, при автоматической сортировке фотографий важно, чтобы в категорию собак не попадало слишком много изображений кошек. Recall важнее, когда необходимо обнаружить как можно больше животных определённого класса, например при поиске потерявшихся собак в большом архиве фотографий.
